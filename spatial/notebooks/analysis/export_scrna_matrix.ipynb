```r
library(ggplot2)
library(SeuratObject)
library(Seurat)
library(presto)
library(dplyr)
library(SeuratDisk)
library(Matrix)
```

```r
sc <-readRDS('data/spatial/processed_data/scRNA.rds')
sc
```

```r
mat <- GetAssayData(sc, assay = 'RNA', slot = 'counts')
writeMM(mat, file = 'data/spatial/processed_data/counts.mtx')
write.csv(rownames(mat),
          'data/spatial/processed_data/gene.csv',
          quote = FALSE,
          row.names = FALSE,
          col.names = FALSE)
metadata <- data.frame(
  donor_id = sc$donor_id,
  cell_type = sc$annotation_level_1
)
write.csv(metadata,
          'data/spatial/processed_data/metadata.csv',
          quote = FALSE,
          row.names = FALSE)
```
