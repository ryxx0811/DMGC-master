```python
import numpy as np
import pandas as pd
import os
```

```python

data_dir = "data/spatial/R_score/BCLL-9-T/"

df_dict = {}
for file in os.listdir(data_dir):
    if file.endswith(".csv"):
        ct = os.path.splitext(file)[0] 
        df = pd.read_csv(os.path.join(data_dir, file), index_col=0)  
        df_dict[ct] = df



df_dict_r4 = {}
for ct, df in df_dict.items():
    other_cts = [k for k in df_dict.keys() if k != ct]
    other_r3 = pd.concat([df_dict[o]['delta_R2'] for o in other_cts], axis=1)
    median_other_r3 = other_r3.median(axis=1)
    df_copy = df.copy()
    df_copy['R4'] = df_copy['delta_R2'] - median_other_r3
    df_dict_r4[ct] = df_copy



output_dir = "data/spatial/R_score/BCLL-9-T/"
for ct, df in df_dict_r4.items():
    df.to_csv(os.path.join(output_dir, f"{ct}.csv"))
'''

all_df = pd.concat(
    [df.assign(celltype=ct) for ct, df in df_dict_r4.items()],
    axis=0
)
all_df.reset_index(inplace=True)
all_df.rename(columns={'index': 'gene'}, inplace=True)
all_df.to_csv(os.path.join(output_dir, "all_celltypes_R4.csv"), index=False)
'''
```
