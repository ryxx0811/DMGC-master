```python
import pandas as pd
import torch
import anndata
import numpy as np
from scipy.io import mmread
from scipy.sparse import csr_matrix
```

```python
X = mmread("data/spatial/processed_data/counts.mtx").T 
X = csr_matrix(X)
genes = pd.read_csv("data/spatial/processed_data/gene.csv")['x']

metadata = pd.read_csv("data/spatial/processed_data/metadata.csv")
donor_id = metadata['donor_id']
cell_type = metadata['cell_type']

adata = anndata.AnnData(X=X)
adata.var_names = genes
adata.obs['donor_id'] = donor_id.values
adata.obs['cell_type'] = cell_type.values
adata
```
