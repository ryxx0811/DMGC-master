```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def topk(sim_matrix, k):
    N = sim_matrix.shape[0]
    adj = np.zeros_like(sim_matrix)

    for i in range(N):
        row = sim_matrix[i].copy()
        row[i] = -np.inf 
        topk_idx = np.argpartition(-row, k)[:k]
        adj[i, topk_idx] = sim_matrix[i, topk_idx]

    adj = np.maximum(adj, adj.T)
    return adj
```

```python
GCBC = torch.load('data/spatial/graph/GCBC.pt',weights_only=False)
FDC = torch.load('data/spatial/graph/FDC.pt',weights_only=False)
epithelial = torch.load('data/spatial/graph/epithelial.pt',weights_only=False)
```

```python
GCBC = topk(GCBC,50)
FDC = topk(FDC,50)
epithelial = topk(epithelial,50)
```

```python
torch.save(GCBC,'data/spatial/graph/GCBC_topk.pt')
torch.save(FDC,'data/spatial/graph/FDC_topk.pt')
torch.save(epithelial,'data/spatial/graph/epithelial_topk.pt')
```

```python
chaffin = torch.load('reheatHeart/data/Chaffin_2022.pt')
kuppe = torch.load('reheatHeart/data/Kuppe_2022.pt')
koenig = torch.load('reheatHeart/data/Koenig_2022.pt')
reichart = torch.load('reheatHeart/data/Reichart_2022.pt')
simonson = torch.load('reheatHeart/data/Simonson_2023.pt')
```
